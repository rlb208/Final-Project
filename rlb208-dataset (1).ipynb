```python
from pathlib import Path
import pandas as pd
file_path = Path('Major Arrests Dataset.csv')
df = pd.read_csv(file_path)
sorted_df = df.sort_values(by='overall_crime_rate')[['neighborhood','overall_crime_rate']]
print(sorted_df)
```

*__Harpen Hilltop is the former name of Summer Hill__*

```python
import matplotlib.pyplot as plt
top_10 = sorted_df.head(10)

plt.figure(figsize=(10, 6))
plt.barh(top_10['neighborhood'], top_10['overall_crime_rate'], color='crimson')
plt.gca().invert_yaxis()

plt.xlabel('Overall Crime Rate')
plt.ylabel('Neighborhood')
plt.title('Top 10 Neighborhoods by Overall Crime Rate')
plt.tight_layout()
plt.show()
```
